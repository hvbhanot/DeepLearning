# 04 - Models as Classes (`nn.Module`)

`nn.Sequential` only handles straight pipelines. Every model past this point in
the course - CNNs with skip connections, autoencoders, GANs - subclasses
`nn.Module` instead, which is the real PyTorch idiom.

**The contract**
1. `__init__` creates the layers and calls `super().__init__()` first (skip that
   call and PyTorch cannot register parameters).
2. Attributes that are layers are auto-registered, so `model.parameters()` finds
   them and the optimizer can train them.
3. `forward(x)` defines the data flow. You call `model(x)`, never
   `model.forward(x)` - the former also runs hooks and train/eval bookkeeping.

**`nn.ReLU()` vs `F.relu()`:** the first is a layer object (for `Sequential`),
the second a plain function (for `forward`). Same math; activations have no
weights, so neither is 'more correct'.

```python
# torch.nn holds layer CLASSES; torch.nn.functional holds the same operations
# as plain FUNCTIONS. Use nn.* for anything with weights, F.* for activations.
import pandas as pd
import numpy as np
import torch
import torch.nn as nn

import torch.nn.functional as F
```

```python
# Same network as the Sequential version, written as a class.
# __init__ declares the parts; forward wires them together. Because forward is
# ordinary Python you can branch, loop, reuse a layer twice, or return several
# tensors - none of which Sequential can express.
class Model(nn.Module):
  def __init__(self):
    super().__init__()

    # Input Layer
    self.input_layer = nn.Linear(2,1)

    # Output Layer
    self.output_layer = nn.Linear(1,1)


  def forward(self,x):

   x = self.input_layer(x)

   x = F.relu(x)

   x = self.output_layer(x)
   x = torch.sigmoid(x)

   return x
```

```python
# Instantiating runs __init__, which creates and randomly initialises weights.
ANN_classify = Model()
```

```python
# printing a Module shows the registered submodules - a quick check that every
# layer you meant to create actually exists and has the shape you expect.
ANN_classify
```
