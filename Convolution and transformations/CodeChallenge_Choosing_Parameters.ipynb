The formula you will use in every CNN you ever build:

    out = floor( (in + 2*padding - kernel) / stride ) + 1

**Why it matters.** A CNN ends by flattening its feature maps into a `Linear`
layer, and that layer's input size has to be exactly right. Get the arithmetic
wrong and you get a shape-mismatch error at the boundary - the single most common
error when writing a CNN from scratch.

**The three useful special cases**
- `padding = kernel//2`, `stride = 1`, odd kernel -> output size equals input size
  ('same' convolution).
- `stride = 2` -> output size roughly halves. This is how CNNs downsample without
  pooling.
- `padding = 0` -> output shrinks by `kernel - 1` each layer, which adds up fast
  in a deep stack.

**The escape hatch:** if the arithmetic gets tedious, run one dummy tensor through
the layers and print the shape - which is exactly what this notebook does to check
the formula.

```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt

import torch
import torch.nn as nn
import torch.nn.functional as F
from torch.utils.data import DataLoader, TensorDataset

from sklearn.preprocessing import MinMaxScaler, StandardScaler
from sklearn.model_selection import train_test_split

!pip install torchvision
import torchvision
import torchvision.transforms as transforms
```

```python
# The challenge: turn a (3,64,64) image into (10,28,28).
# out_channels=10 is free - it is just the number of kernels. The spatial size
# is the constrained part, solved here with kernel=9, stride=2, padding=0.
print("Make the input image of [3,64,64] to convole to [10,28,28] : ")

img_size = (3,64,64)

input_chan = 3
ouput_chan = 10

stride = 2
padding = 0
kernal = 9

img = torch.rand(1,img_size[0],img_size[1],img_size[2])

c = nn.Conv2d(in_channels=input_chan, out_channels= ouput_chan , stride=stride, padding=padding, kernel_size=kernal)

img.shape
```

```python
# Theory against practice.
# floor((64 + 0 - 9)/2) + 1 = floor(27.5) + 1 = 28. Confirmed by running a real
# tensor through the layer.
# The dummy-tensor trick is the practical way to size the flatten->Linear
# boundary in any CNN: build the conv stack, push one input through it, print
# the shape, and use that number.
out_h = np.floor( (img_size[1] + 2*(padding) - kernal) / stride ) + 1
out_w = np.floor( (img_size[2] + 2*(padding) - kernal) / stride ) + 1


output = c(img)

print(f"Theruretical Size: ({out_h} , {out_w})")
print(f'Eperical Size : {output.shape}')


```
