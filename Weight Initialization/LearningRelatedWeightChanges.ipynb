**Section:** Weight Initialization

A stub. The intended topic is tracking how much each layer's weights change over
training - typically by recording the Frobenius norm of `weight - weight_initial`
per layer, per epoch, and plotting it.

**What that plot tells you**
- Early layers usually change *less* than later ones, especially in a deep
  network - gradients shrink on the way back.
- A layer whose weights never move is dead: its gradient is vanishing, or its
  learning rate is far too small for it.
- Weights that change explosively signal a learning rate that is too high.
